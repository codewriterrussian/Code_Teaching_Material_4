## 2. 第 1 題：商品總價

商店購買一種商品。請輸入商品的**單價**與**數量**，計算總價。

### 輸入

兩個整數：

```text
price quantity
```

### 輸出

```text
Total = 答案
```

### 範例輸入

```text
35 8
```

### 範例輸出

```text
Total = 280
```


### C 解法


````python
%%c -n 1

#include <stdio.h>

int main(void){
    int price, quantity;
    int total;

    scanf("%d %d", &price, &quantity);

    total = price * quantity;

    printf("Total = %d\n", total);

    return 0;
}
````

### C++ 解法


````python
%%cpp -n 1

#include <iostream>
using namespace std;

int main(){
    int price, quantity;
    int total;

    cin >> price >> quantity;

    total = price * quantity;

    cout << "Total = " << total << endl;

    return 0;
}
````

## 3. 第 2 題：平均成績

輸入三次考試成績，計算三次成績的平均值。

平均值請輸出到**小數點後兩位**。

### 輸入

```text
score1 score2 score3
```

### 輸出

```text
Average = 答案
```

### 範例輸入

```text
80 91 76
```

### 範例輸出

```text
Average = 82.33
```


### C 解法


````python
%%c -n 1

#include <stdio.h>

int main(void){
    int score1, score2, score3;
    double average;

    scanf("%d %d %d", &score1, &score2, &score3);

    average = (score1 + score2 + score3) / 3.0;

    printf("Average = %.2f\n", average);

    return 0;
}
````

### C++ 解法


````python
%%cpp -n 1

#include <iostream>
#include <iomanip>
using namespace std;

int main(){
    int score1, score2, score3;
    double average;

    cin >> score1 >> score2 >> score3;

    average = (score1 + score2 + score3) / 3.0;

    cout << fixed << setprecision(2);
    cout << "Average = " << average << endl;

    return 0;
}
````

## 4. 第 3 題：秒數轉換

輸入一個總秒數，將它轉換成：

```text
小時 分鐘 秒
```

### 輸入

一個非負整數：

```text
seconds
```

### 輸出

```text
H = 小時
M = 分鐘
S = 秒
```

### 範例輸入

```text
7384
```

### 範例輸出

```text
H = 2
M = 3
S = 4
```


### C 解法


````python
%%c -n 1

#include <stdio.h>

int main(void){
    int totalsec;
    int hrs, mins, secs;

    // 輸入
    scanf("%d", &totalsec);

    // 運算 (% / )
    hrs = totalsec / 3600;
    mins = (totalsec % 3600) / 60;
    secs = totalsec % 60;

    // 輸出
    printf("H = %d\n", hrs);
    printf("M = %d\n", mins);
    printf("S = %d\n", secs);

    return 0;
}
````

### C++ 解法


````python
%%cpp -n 1

#include <iostream>
using namespace std;

int main(void){
    int totalsec;
    int hrs, mins, secs;

    // 輸入
    cin >> totalsec;

    // 運算 (% / )
    hrs = totalsec / 3600;
    mins = (totalsec % 3600) / 60;
    secs = totalsec % 60;

    // 輸出
    cout << "H = " << hrs << endl;
    cout << "M = " << mins << endl;
    cout << "S = " << secs << endl;

    return 0;
}
````
