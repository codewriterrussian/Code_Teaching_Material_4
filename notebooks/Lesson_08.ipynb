```python
# factorial -> 階層 -> 9!: 9 * 8 * 7 * 6 * 5 * 4 * 3 * 2 * 1 * 1
# f(9) = 9 * f(8)
def factorial(n):
    if n == 1:
        return 1
    else:
        return n * factorial(n - 1)
```
