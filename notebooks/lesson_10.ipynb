```python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int x = 10;
    int *p = &x;

    cout << &x << endl;
    cout << p << endl;
    cout << *p <<endl;

    return 0;
}
```

```python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int x = 10;
    int *p = &x;

    *p = 50;

    cout << x << endl;

    return 0;
}
```
