### 練習：將整數顯示至小數點後兩位

程式會持續要求使用者輸入，直到輸入內容為非負整數，再以兩位小數顯示結果。


````python
number_text = input("請輸入一個非負整數：")

while not number_text.isdigit():
    print(
        f"You typed {number_text!r}. "
        "This is not a non-negative integer."
    )
    number_text = input("請重新輸入：")

number = int(number_text)
print(f"{number:.2f}")
````

## 4. BMI 計算與分類

BMI 計算公式：

\[
\text{BMI} = \frac{\text{體重（公斤）}}{\text{身高（公尺）}^2}
\]

| BMI 範圍 | 分類 |
|---|---|
| BMI < 18.5 | 體重過輕 |
| 18.5 ≤ BMI < 24 | 正常範圍 |
| 24 ≤ BMI < 27 | 過重 |
| 27 ≤ BMI < 30 | 輕度肥胖 |
| 30 ≤ BMI < 35 | 中度肥胖 |
| BMI ≥ 35 | 重度肥胖 |

程式步驟：

1. 輸入體重與身高。
2. 計算 BMI。
3. 將 BMI 顯示至小數點後一位。
4. 顯示 BMI 所屬分類。


````python
weight = float(input("輸入體重（公斤）："))
height = float(input("輸入身高（公尺）："))

bmi = weight / (height ** 2)

if bmi < 18.5:
    result = "體重過輕"
elif bmi < 24:
    result = "正常範圍"
elif bmi < 27:
    result = "過重"
elif bmi < 30:
    result = "肥胖－輕度"
elif bmi < 35:
    result = "肥胖－中度"
else:
    result = "肥胖－重度"

print(f"BMI：{bmi:.1f}")
print(f"結果：{result}")
````

## 5. 五位數拆解

以下函式假設輸入為一個五位正整數，並回傳：

- 五個位數
- 所有位數的總和


### 方法一：使用整除與餘數

````python
def q1_five_digit_profile(number):
    ten_thousands = number // 10000
    thousands = (number // 1000) % 10
    hundreds = (number // 100) % 10
    tens = (number // 10) % 10
    ones = number % 10

    digit_sum = (
        ten_thousands
        + thousands
        + hundreds
        + tens
        + ones
    )

    return (
        ten_thousands,
        thousands,
        hundreds,
        tens,
        ones,
        digit_sum,
    )


print(q1_five_digit_profile(12345))
````

### 方法二：使用迴圈

- `% 10` 取得個位數。
- `// 10` 移除個位數。
- 因為數字是由右至左取得，所以最後需要反轉串列。


````python
def q1_five_digit_profile_iterative(number):
    digits = []

    for _ in range(5):
        digits.append(number % 10)
        number //= 10

    digits.reverse()
    return (*digits, sum(digits))


print(q1_five_digit_profile_iterative(12345))
````

## 6. 反轉五位數

例如：

```text
12345 → 54321
```

前導零在整數中不會保留，例如 `00100` 會以整數 `100` 表示。


````python
def q2_reverse_five_digits(number):
    ten_thousands = number // 10000
    thousands = (number // 1000) % 10
    hundreds = (number // 100) % 10
    tens = (number // 10) % 10
    ones = number % 10

    return (
        ones * 10000
        + tens * 1000
        + hundreds * 100
        + thousands * 10
        + ten_thousands
    )


print(q2_reverse_five_digits(12345))
````

## 7. 三位數向左旋轉

將百位數移到個位數位置：

```text
123 → 231
```


````python
def q3_rotate_left_three_digits(number):
    hundreds = number // 100
    remaining = number % 100

    return remaining * 10 + hundreds


print(q3_rotate_left_three_digits(123))
````
