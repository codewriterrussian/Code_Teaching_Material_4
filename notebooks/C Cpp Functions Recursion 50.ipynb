````python
# === C / C++ 自動檢查工具：請先執行這個 cell，不要修改 ===
import base64
import json
import os
import shutil
import subprocess
import tempfile
from IPython.core.magic import register_cell_magic

_TEST_CASES = json.loads(
    base64.b64decode("eyJRMSI6W3siaW5wdXQiOiI0XG4iLCJleHBlY3RlZCI6IjdcbiJ9LHsiaW5wdXQiOiItMlxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiMFxuIiwiZXhwZWN0ZWQiOiIzXG4ifV0sIlEyIjpbeyJpbnB1dCI6IjVcbiIsImV4cGVjdGVkIjoiMjVcbiJ9LHsiaW5wdXQiOiItN1xuIiwiZXhwZWN0ZWQiOiI0OVxuIn0seyJpbnB1dCI6IjBcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMyI6W3siaW5wdXQiOiItOFxuIiwiZXhwZWN0ZWQiOiI4XG4ifSx7ImlucHV0IjoiMTJcbiIsImV4cGVjdGVkIjoiMTJcbiJ9LHsiaW5wdXQiOiIwXG4iLCJleHBlY3RlZCI6IjBcbiJ9XSwiUTQiOlt7ImlucHV0IjoiMTRcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjlcbiIsImV4cGVjdGVkIjoiMFxuIn0seyJpbnB1dCI6IjBcbiIsImV4cGVjdGVkIjoiMVxuIn1dLCJRNSI6W3siaW5wdXQiOiIxMFxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiLTNcbiIsImV4cGVjdGVkIjoiLTFcbiJ9LHsiaW5wdXQiOiIwXG4iLCJleHBlY3RlZCI6IjBcbiJ9XSwiUTYiOlt7ImlucHV0IjoiMyA1XG4iLCJleHBlY3RlZCI6IjVcbiJ9LHsiaW5wdXQiOiI4IDJcbiIsImV4cGVjdGVkIjoiOFxuIn0seyJpbnB1dCI6Ii00IC03XG4iLCJleHBlY3RlZCI6Ii00XG4ifV0sIlE3IjpbeyJpbnB1dCI6IjMgNVxuIiwiZXhwZWN0ZWQiOiIzXG4ifSx7ImlucHV0IjoiOCAyXG4iLCJleHBlY3RlZCI6IjJcbiJ9LHsiaW5wdXQiOiItNCAtN1xuIiwiZXhwZWN0ZWQiOiItN1xuIn1dLCJROCI6W3siaW5wdXQiOiIxNSAxMCAyMFxuIiwiZXhwZWN0ZWQiOiIxNVxuIn0seyJpbnB1dCI6IjMgMTAgMjBcbiIsImV4cGVjdGVkIjoiMTBcbiJ9LHsiaW5wdXQiOiIyNSAxMCAyMFxuIiwiZXhwZWN0ZWQiOiIyMFxuIn1dLCJROSI6W3siaW5wdXQiOiIxNSAxMCAyMFxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiMTAgMTAgMjBcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjIxIDEwIDIwXG4iLCJleHBlY3RlZCI6IjBcbiJ9XSwiUTEwIjpbeyJpbnB1dCI6IjIwIDVcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjIwIDZcbiIsImV4cGVjdGVkIjoiMFxuIn0seyJpbnB1dCI6IjIwIDBcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMTEiOlt7ImlucHV0IjoiMyA5IDVcbiIsImV4cGVjdGVkIjoiOVxuIn0seyJpbnB1dCI6IjEwIC0yIDhcbiIsImV4cGVjdGVkIjoiMTBcbiJ9LHsiaW5wdXQiOiItNCAtNyAtMVxuIiwiZXhwZWN0ZWQiOiItMVxuIn1dLCJRMTIiOlt7ImlucHV0IjoiMyA5IDVcbiIsImV4cGVjdGVkIjoiM1xuIn0seyJpbnB1dCI6IjEwIC0yIDhcbiIsImV4cGVjdGVkIjoiLTJcbiJ9LHsiaW5wdXQiOiItNCAtNyAtMVxuIiwiZXhwZWN0ZWQiOiItN1xuIn1dLCJRMTMiOlt7ImlucHV0IjoiMyAtMSAyXG4iLCJleHBlY3RlZCI6IjJcbiJ9LHsiaW5wdXQiOiIwIDAgMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiLTEgLTIgLTNcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMTQiOlt7ImlucHV0IjoiNVxuIiwiZXhwZWN0ZWQiOiIxNVxuIn0seyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjEwMFxuIiwiZXhwZWN0ZWQiOiI1MDUwXG4ifV0sIlExNSI6W3siaW5wdXQiOiI1XG4iLCJleHBlY3RlZCI6IjEyMFxuIn0seyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjEwXG4iLCJleHBlY3RlZCI6IjM2Mjg4MDBcbiJ9XSwiUTE2IjpbeyJpbnB1dCI6IjEyXG4iLCJleHBlY3RlZCI6IjZcbiJ9LHsiaW5wdXQiOiIxM1xuIiwiZXhwZWN0ZWQiOiIyXG4ifSx7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifV0sIlExNyI6W3siaW5wdXQiOiIyOVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiMjVcbiIsImV4cGVjdGVkIjoiMFxuIn0seyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMTgiOlt7ImlucHV0IjoiNDggMThcbiIsImV4cGVjdGVkIjoiNlxuIn0seyJpbnB1dCI6IjE3IDVcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjQyIDU2XG4iLCJleHBlY3RlZCI6IjE0XG4ifV0sIlExOSI6W3siaW5wdXQiOiI1ODMxXG4iLCJleHBlY3RlZCI6IjRcbiJ9LHsiaW5wdXQiOiIwXG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiIxMDAwMDBcbiIsImV4cGVjdGVkIjoiNlxuIn1dLCJRMjAiOlt7ImlucHV0IjoiNTgzMVxuIiwiZXhwZWN0ZWQiOiIxN1xuIn0seyJpbnB1dCI6IjkwMDA5XG4iLCJleHBlY3RlZCI6IjE4XG4ifSx7ImlucHV0IjoiMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifV0sIlEyMSI6W3siaW5wdXQiOiI1XG4iLCJleHBlY3RlZCI6IjEwMCA1XG4ifSx7ImlucHV0IjoiLTNcbiIsImV4cGVjdGVkIjoiMTAwIC0zXG4ifSx7ImlucHV0IjoiMTAwXG4iLCJleHBlY3RlZCI6IjEwMCAxMDBcbiJ9XSwiUTIyIjpbeyJpbnB1dCI6IiMgNVxuIiwiZXhwZWN0ZWQiOiIjIyMjI1xuIn0seyJpbnB1dCI6IkEgMVxuIiwiZXhwZWN0ZWQiOiJBXG4ifSx7ImlucHV0IjoiKiAzXG4iLCJleHBlY3RlZCI6IioqKlxuIn1dLCJRMjMiOlt7ImlucHV0IjoiIyAyIDRcbiIsImV4cGVjdGVkIjoiIyMjI1xuIyMjI1xuIn0seyJpbnB1dCI6IkEgMSAzXG4iLCJleHBlY3RlZCI6IkFBQVxuIn0seyJpbnB1dCI6IiogMyAxXG4iLCJleHBlY3RlZCI6IipcbipcbipcbiJ9XSwiUTI0IjpbeyJpbnB1dCI6IiMgNFxuIiwiZXhwZWN0ZWQiOiIjXG4jI1xuIyMjXG4jIyMjXG4ifSx7ImlucHV0IjoiQSAxXG4iLCJleHBlY3RlZCI6IkFcbiJ9LHsiaW5wdXQiOiIqIDNcbiIsImV4cGVjdGVkIjoiKlxuKipcbioqKlxuIn1dLCJRMjUiOlt7ImlucHV0IjoiIyA0XG4iLCJleHBlY3RlZCI6IiMjIyNcbiMjI1xuIyNcbiNcbiJ9LHsiaW5wdXQiOiJBIDJcbiIsImV4cGVjdGVkIjoiQUFcbkFcbiJ9LHsiaW5wdXQiOiIqIDFcbiIsImV4cGVjdGVkIjoiKlxuIn1dLCJRMjYiOlt7ImlucHV0IjoiNFxuIiwiZXhwZWN0ZWQiOiIxXG4yXG4zXG40XG4ifSx7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiNlxuIiwiZXhwZWN0ZWQiOiIxXG4yXG4zXG40XG41XG42XG4ifV0sIlEyNyI6W3siaW5wdXQiOiI0XG4iLCJleHBlY3RlZCI6IjRcbjNcbjJcbjFcbiJ9LHsiaW5wdXQiOiIxXG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiI2XG4iLCJleHBlY3RlZCI6IjZcbjVcbjRcbjNcbjJcbjFcbiJ9XSwiUTI4IjpbeyJpbnB1dCI6IjNcbiIsImV4cGVjdGVkIjoiRW50ZXIgMVxuRW50ZXIgMlxuRW50ZXIgM1xuTGVhdmUgM1xuTGVhdmUgMlxuTGVhdmUgMVxuIn0seyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiRW50ZXIgMVxuTGVhdmUgMVxuIn0seyJpbnB1dCI6IjRcbiIsImV4cGVjdGVkIjoiRW50ZXIgMVxuRW50ZXIgMlxuRW50ZXIgM1xuRW50ZXIgNFxuTGVhdmUgNFxuTGVhdmUgM1xuTGVhdmUgMlxuTGVhdmUgMVxuIn1dLCJRMjkiOlt7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiNFxuIiwiZXhwZWN0ZWQiOiIxMFxuIn0seyJpbnB1dCI6IjEwXG4iLCJleHBlY3RlZCI6IjU1XG4ifV0sIlEzMCI6W3siaW5wdXQiOiIxXG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiI1XG4iLCJleHBlY3RlZCI6IjEyMFxuIn0seyJpbnB1dCI6IjEwXG4iLCJleHBlY3RlZCI6IjM2Mjg4MDBcbiJ9XSwiUTMxIjpbeyJpbnB1dCI6IjIgNVxuIiwiZXhwZWN0ZWQiOiIzMlxuIn0seyJpbnB1dCI6Ii0zIDNcbiIsImV4cGVjdGVkIjoiLTI3XG4ifSx7ImlucHV0IjoiNyAwXG4iLCJleHBlY3RlZCI6IjFcbiJ9XSwiUTMyIjpbeyJpbnB1dCI6IjU4MzFcbiIsImV4cGVjdGVkIjoiMTdcbiJ9LHsiaW5wdXQiOiI5MDAwOVxuIiwiZXhwZWN0ZWQiOiIxOFxuIn0seyJpbnB1dCI6IjBcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMzMiOlt7ImlucHV0IjoiNTgzMVxuIiwiZXhwZWN0ZWQiOiI0XG4ifSx7ImlucHV0IjoiMFxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiMTAwMDAwXG4iLCJleHBlY3RlZCI6IjZcbiJ9XSwiUTM0IjpbeyJpbnB1dCI6IjUzOFxuIiwiZXhwZWN0ZWQiOiI1IDMgOFxuIn0seyJpbnB1dCI6IjdcbiIsImV4cGVjdGVkIjoiN1xuIn0seyJpbnB1dCI6IjEwMDJcbiIsImV4cGVjdGVkIjoiMSAwIDAgMlxuIn1dLCJRMzUiOlt7ImlucHV0IjoiNTM4XG4iLCJleHBlY3RlZCI6IjggMyA1XG4ifSx7ImlucHV0IjoiN1xuIiwiZXhwZWN0ZWQiOiI3XG4ifSx7ImlucHV0IjoiMTAwMlxuIiwiZXhwZWN0ZWQiOiIyIDAgMCAxXG4ifV0sIlEzNiI6W3siaW5wdXQiOiIjIDVcbiIsImV4cGVjdGVkIjoiIyMjIyNcbiJ9LHsiaW5wdXQiOiJBIDFcbiIsImV4cGVjdGVkIjoiQVxuIn0seyJpbnB1dCI6IiogM1xuIiwiZXhwZWN0ZWQiOiIqKipcbiJ9XSwiUTM3IjpbeyJpbnB1dCI6IjEwXG4iLCJleHBlY3RlZCI6IjMwXG4ifSx7ImlucHV0IjoiOVxuIiwiZXhwZWN0ZWQiOiIyMFxuIn0seyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMzgiOlt7ImlucHV0IjoiMTBcbiIsImV4cGVjdGVkIjoiMjVcbiJ9LHsiaW5wdXQiOiI5XG4iLCJleHBlY3RlZCI6IjI1XG4ifSx7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifV0sIlEzOSI6W3siaW5wdXQiOiIxMFxuIiwiZXhwZWN0ZWQiOiIxMFxuOFxuNlxuNFxuMlxuIn0seyJpbnB1dCI6IjlcbiIsImV4cGVjdGVkIjoiOVxuN1xuNVxuM1xuMVxuIn0seyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMVxuIn1dLCJRNDAiOlt7ImlucHV0IjoiNDggMThcbiIsImV4cGVjdGVkIjoiNlxuIn0seyJpbnB1dCI6IjE3IDVcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjQyIDU2XG4iLCJleHBlY3RlZCI6IjE0XG4ifV0sIlE0MSI6W3siaW5wdXQiOiI3IDVcbiIsImV4cGVjdGVkIjoiMzVcbiJ9LHsiaW5wdXQiOiItMyA0XG4iLCJleHBlY3RlZCI6Ii0xMlxuIn0seyJpbnB1dCI6IjkgMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifV0sIlE0MiI6W3siaW5wdXQiOiIzXG4iLCJleHBlY3RlZCI6IjE0XG4ifSx7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiNVxuIiwiZXhwZWN0ZWQiOiI1NVxuIn1dLCJRNDMiOlt7ImlucHV0IjoiNTA1MDUwIDVcbiIsImV4cGVjdGVkIjoiM1xuIn0seyJpbnB1dCI6IjEwMDAgMFxuIiwiZXhwZWN0ZWQiOiIzXG4ifSx7ImlucHV0IjoiNyA3XG4iLCJleHBlY3RlZCI6IjFcbiJ9XSwiUTQ0IjpbeyJpbnB1dCI6IjU4MzFcbiIsImV4cGVjdGVkIjoiOFxuIn0seyJpbnB1dCI6IjkwMDA5XG4iLCJleHBlY3RlZCI6IjlcbiJ9LHsiaW5wdXQiOiI3XG4iLCJleHBlY3RlZCI6IjdcbiJ9XSwiUTQ1IjpbeyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjVcbiIsImV4cGVjdGVkIjoiOFxuIn0seyJpbnB1dCI6IjEwXG4iLCJleHBlY3RlZCI6Ijg5XG4ifV0sIlE0NiI6W3siaW5wdXQiOiIxXG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiI1XG4iLCJleHBlY3RlZCI6IjEzXG4ifSx7ImlucHV0IjoiOFxuIiwiZXhwZWN0ZWQiOiI4MVxuIn1dLCJRNDciOlt7ImlucHV0IjoiMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifSx7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiMTBcbiIsImV4cGVjdGVkIjoiNTVcbiJ9XSwiUTQ4IjpbeyJpbnB1dCI6IjFcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjRcbiIsImV4cGVjdGVkIjoiMTVcbiJ9LHsiaW5wdXQiOiIxMFxuIiwiZXhwZWN0ZWQiOiIxMDIzXG4ifV0sIlE0OSI6W3siaW5wdXQiOiIwXG4iLCJleHBlY3RlZCI6IjBcbiJ9LHsiaW5wdXQiOiI1XG4iLCJleHBlY3RlZCI6IjEwMVxuIn0seyJpbnB1dCI6IjE4XG4iLCJleHBlY3RlZCI6IjEwMDEwXG4ifV0sIlE1MCI6W3siaW5wdXQiOiI1MzgxXG4iLCJleHBlY3RlZCI6IjE4MzVcbiJ9LHsiaW5wdXQiOiIxMjAwXG4iLCJleHBlY3RlZCI6IjIxXG4ifSx7ImlucHV0IjoiMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifV19").decode("utf-8")
)

_ANSWERS = {}
_RESULTS = {}

def _normalise_output(text):
    lines = [line.rstrip() for line in text.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _find_compiler(language):
    if language == "c":
        names = ["gcc", "clang", "cc"]
    else:
        names = ["g++", "clang++", "c++"]

    for name in names:
        path = shutil.which(name)
        if path:
            return path
    return None

def _compile_and_run(source, input_text, language):
    compiler = _find_compiler(language)

    if compiler is None:
        if language == "c":
            return False, "", "找不到 C 編譯器（gcc / clang / cc）。"
        return False, "", "找不到 C++ 編譯器（g++ / clang++ / c++）。"

    with tempfile.TemporaryDirectory() as td:
        if language == "c":
            src = os.path.join(td, "answer.c")
            compile_cmd = [compiler, "-std=c11", src, "-o", os.path.join(td, "answer")]
        else:
            src = os.path.join(td, "answer.cpp")
            compile_cmd = [compiler, "-std=c++17", src, "-o", os.path.join(td, "answer")]

        exe = os.path.join(td, "answer")

        with open(src, "w", encoding="utf-8") as f:
            f.write(source)

        compile_result = subprocess.run(
            compile_cmd,
            capture_output=True,
            text=True,
        )

        if compile_result.returncode != 0:
            return False, "", compile_result.stderr

        try:
            run_result = subprocess.run(
                [exe],
                input=input_text,
                capture_output=True,
                text=True,
                timeout=5,
            )
        except subprocess.TimeoutExpired:
            return False, "", "程式執行超過 5 秒，請檢查是否有無限迴圈或無限遞迴。"

        if run_result.returncode != 0:
            err = run_result.stderr or f"程式結束碼：{run_result.returncode}"
            return False, run_result.stdout, err

        return True, run_result.stdout, ""

def _save_answer(qid, language, cell):
    qid = qid.strip().upper()
    if qid not in _TEST_CASES:
        print(f"❌ 未知題號：{qid}")
        return

    _ANSWERS[qid] = {"language": language, "source": cell}
    label = "C" if language == "c" else "C++"
    print(f"已儲存 {qid} 的 {label} 作答。請執行下方檢查 cell。")

@register_cell_magic
def c_answer(line, cell):
    _save_answer(line, "c", cell)

@register_cell_magic
def cpp_answer(line, cell):
    _save_answer(line, "cpp", cell)

def _run_check(qid):
    qid = qid.upper()

    if qid not in _ANSWERS:
        print(f"❌ {qid}：尚未執行 C 或 C++ 作答 cell")
        _RESULTS[qid] = False
        return

    answer = _ANSWERS[qid]
    language = answer["language"]
    source = answer["source"]
    cases = _TEST_CASES[qid]
    passed = 0

    for case_number, case in enumerate(cases, start=1):
        ok_run, stdout, error = _compile_and_run(
            source,
            case["input"],
            language,
        )

        if not ok_run:
            label = "C" if language == "c" else "C++"
            print(f"❌ {qid}：{label} 編譯或執行失敗")
            print(error)
            _RESULTS[qid] = False
            return

        actual = _normalise_output(stdout)
        expected = _normalise_output(case["expected"])

        if actual == expected:
            passed += 1
        else:
            print(f"❌ {qid} 測試 {case_number} 未通過")
            print("測試輸入：")
            print(case["input"], end="")
            print("你的輸出：")
            print(stdout if stdout else "<沒有輸出>")
            print("預期輸出：")
            print(case["expected"], end="")
            print()

    success = passed == len(cases)
    _RESULTS[qid] = success
    label = "C" if language == "c" else "C++"

    if success:
        print(f"✅ {qid} 全部通過（{passed} / {len(cases)}，{label}）")
    else:
        print(f"❌ {qid}：通過 {passed} / {len(cases)} 個測試（{label}）")

def _final_report():
    passed = sum(_RESULTS.get(f"Q{i}", False) for i in range(1, 51))
    untouched = [f"Q{i}" for i in range(1, 51) if f"Q{i}" not in _RESULTS]

    print(f"總成績：{passed} / 50 題通過")
    if untouched:
        print("尚未執行檢查：" + ", ".join(untouched))

print("C / C++ 自動檢查工具已載入。")
print("每題可選 C 或 C++；最後執行的作答 cell 會作為該題答案。")
print("作答後不需要手動輸入測試資料。")
````
