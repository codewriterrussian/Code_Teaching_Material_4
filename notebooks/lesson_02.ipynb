## 1. 使用索引走訪串列

`range(len(lst))` 會產生串列的有效索引。

再使用 `lst[index]` 取得每個位置的元素。

```python
lst = [1, 2, 3, 4]

for index in range(len(lst)):
    print(lst[index])
```

## 2. 直接走訪串列元素

若不需要索引，可以直接讓迴圈變數依序取得串列中的元素。

這種寫法通常比使用索引更簡潔。

```python
for value in lst:
    print(value)
```

## 3. `range()` 與迴圈變數

`range(5)` 依序提供 `0`、`1`、`2`、`3`、`4`。

即使在迴圈內執行 `i += 1`，下一輪的 `i` 仍然會由 `range(5)` 指定，因此輸出不會變成隔一個數字。

```python
for i in range(5):
    print(i)
    i += 1
```

## 4. 在迴圈中修改串列

以下範例會把「下一個位置」的元素加上 `2`。

`% len(lst)` 可以讓最後一個索引回到第一個索引，形成循環效果。

```python
lst = [1, 2, 3, 4]

for index in range(len(lst)):
    next_index = (index + 1) % len(lst)
    lst[next_index] += 2
    print(lst[index])
```

## 5. 使用 `enumerate()`

`enumerate(lst)` 可以在每次迴圈中同時取得：

- `index`：目前元素的索引
- `element`：目前位置的元素

此範例仍然會修改下一個位置的值。

```python
lst = [1, 2, 3, 4]

for index, element in enumerate(lst):
    next_index = (index + 1) % len(lst)
    lst[next_index] += 2
    print(index, element)
```

## 6. 使用 `continue` 跳過一次迴圈

當 `i == 4` 時，`continue` 會立即結束該次迴圈，直接進入下一次迴圈。

因此數字 `4` 不會被印出。

```python
for i in range(10):
    if i == 4:
        continue

    print(i)
```

## 7. 將總秒數拆分成天、時、分、秒

計算順序如下：

1. 使用整除 `//` 取得完整單位數量。
2. 使用餘數 `%` 保留尚未分配的秒數。
3. 依序計算天、時、分、秒。

```python
def q4_seconds_report(total_seconds):
    # 依序拆成天、時、分、秒
    days = total_seconds // 86_400
    total_seconds %= 86_400

    hours = total_seconds // 3_600
    total_seconds %= 3_600

    minutes = total_seconds // 60
    seconds = total_seconds % 60

    return days, hours, minutes, seconds
```

## 8. 計算距離指定時間還有幾分鐘

先把開始時間與結束時間都換算成分鐘。

使用 `% 1440` 可以處理跨日情況，因為一天共有 `1440` 分鐘。

```python
def q5_minutes_until(
    start_hour,
    start_minute,
    end_hour,
    end_minute,
):
    # 先全部換成分鐘，再處理跨日
    start = start_hour * 60 + start_minute
    end = end_hour * 60 + end_minute

    return (end - start) % 1_440
```

## 9. 計算含稅購物總額

函式會先計算三種商品的小計，再乘上含稅倍率。

最後使用 `round(..., 2)` 保留兩位小數。

```python
def q6_taxed_cart_total(
    p1,
    q1,
    p2,
    q2,
    p3,
    q3,
    tax_rate,
):
    subtotal = p1 * q1 + p2 * q2 + p3 * q3
    total = subtotal * (1 + tax_rate)

    return round(total, 2)


print(q6_taxed_cart_total(100, 2, 50, 1, 30, 3, 0.05))
```

## 10. 計算加權成績

成績比例為：

- 考試：50%
- 專題：30%
- 作業：20%

函式會回傳：

1. 加權平均分數
2. 是否達到 `60` 分

```python
def q7_weighted_grade(exam, project, homework):
    # 計算加權平均，並判斷是否達標
    average = round(
        exam * 0.5
        + project * 0.3
        + homework * 0.2,
        2,
    )

    return average, average >= 60
```

## 重點整理

- `range(len(lst))`：使用索引走訪串列。
- `for value in lst`：直接取得串列元素。
- `enumerate(lst)`：同時取得索引與元素。
- `continue`：跳過目前這一次迴圈。
- `//`：取得整除結果。
- `%`：取得餘數，也可用來處理循環索引或跨日時間。
- 函式可以把一組計算步驟包裝成可重複使用的程式。
