## 1. `switch`：可以用整數、字元，但不能直接用字串

原始速記：

```text
switch -> 1, 'a' => "apple" X
```

這是在提醒：

```cpp
switch
```

適合拿來判斷像下面這類值：

```cpp
1
2
3
'a'
'b'
'c'
```

但不能直接寫成用字串：

```cpp
"apple"
```

來當一般 `case` 的值。

---

### 整數可以

```cpp
switch (choice) {
    case 1:
        // ...
        break;

    case 2:
        // ...
        break;
}
```

### 字元也可以

```cpp
switch (letter) {
    case 'a':
        // ...
        break;

    case 'b':
        // ...
        break;
}
```

這裡：

```cpp
'a'
```

是**單一字元**，所以可以用在這種 `switch` 判斷中。


### `"apple"` 為什麼不行？

要注意：

```cpp
'a'
```

和：

```cpp
"apple"
```

不是同一種東西。

```text
'a'
↓
單一字元

"apple"
↓
字串
```

因此這種想法：

```cpp
case "apple":
```

不是這堂課要使用的 `switch` 寫法。

如果要判斷字串內容，通常會改用其他方式，例如 `if`。

### 記憶方式

```text
switch
→ 1        ✓
→ 'a'      ✓
→ "apple"  ✗
```


## 2. `i % 3`：餘數運算

原始速記：

```cpp
i % 3
```

`%` 是**餘數運算子**。

例如：

```text
0 % 3 = 0
1 % 3 = 1
2 % 3 = 2
3 % 3 = 0
4 % 3 = 1
5 % 3 = 2
6 % 3 = 0
```

所以：

```cpp
i % 3
```

的結果會一直在：

```text
0, 1, 2
```

之間循環。


### `i % 3` 常用來做什麼？

如果某件事情每 3 次重複一次，就可以利用：

```cpp
i % 3
```

來判斷目前在第幾個位置。

例如：

```cpp
if (i % 3 == 0) {
    // 第 0 類
}
else if (i % 3 == 1) {
    // 第 1 類
}
else {
    // 第 2 類
}
```

可以想成：

```text
i = 0 → 0
i = 1 → 1
i = 2 → 2
i = 3 → 0
i = 4 → 1
i = 5 → 2
...
```

因此 `%` 很適合拿來處理**週期、循環、分組**。


## 3. `if True:`：這是概念速記，不是 C/C++ 原始寫法

原始 notebook 寫：

```python
if True:
```

這是 **Python 形式的寫法**。

如果這堂課是在教 C/C++，比較接近的 C++ 寫法是：

```cpp
if (true) {
    // ...
}
```

注意差異：

| Python | C++ |
|---|---|
| `True` | `true` |
| `if True:` | `if (true) { ... }` |

所以原始筆記中的：

```python
if True:
```

可以把它當成老師快速表示：

> 「如果這個條件是真的，就執行裡面的程式。」

而不是要學生照抄成 C/C++。


### 在 C 裡怎麼理解？

最基本可以先理解：

```c
if (1) {
    /* 條件成立 */
}
```

在條件判斷中：

```text
0
→ false

非 0
→ true
```

因此：

```c
if (1)
```

代表條件一定成立。

這裡先保留在初學程度，不延伸其他布林型態細節。


## 5. 常見混淆

### 混淆 1：`'a'` 和 `"a"`

```cpp
'a'
```

是單一字元。

```cpp
"a"
```

是字串。

兩者不要混在一起。

---

### 混淆 2：`%` 不是百分比

在 C/C++：

```cpp
7 % 3
```

不是「7 的 3%」。

它是在算：

```text
7 ÷ 3
```

的餘數，所以答案是：

```text
1
```

---

### 混淆 3：C++ 是 `true`，不是 `True`

```cpp
true
```

是 C++ 的寫法。

```python
True
```

是 Python 的寫法。

如果是在 C++ 程式中，不要把大小寫直接照 Python 寫。
