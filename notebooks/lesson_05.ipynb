# 1. 基本程式流程：求兩個整數的和

原題目：

```text
求兩個整數的和

完成：
輸入 first
輸入 second
計算 sum
輸出 sum
```

這是一個很重要的基礎流程：

```text
輸入
 ↓
儲存
 ↓
計算
 ↓
輸出
```

也可以寫成：

```text
first  ─┐
        ├─→ first + second ─→ sum ─→ output
second ─┘
```


## 1.1 先看變數

我們需要三個整數變數：

```c
int first, second, sum;
```

各自負責：

| 變數 | 用途 |
|---|---|
| `first` | 第一個輸入的整數 |
| `second` | 第二個輸入的整數 |
| `sum` | 儲存 `first + second` 的結果 |

例如輸入：

```text
7
12
```

那麼：

```text
first  = 7
second = 12
sum    = 19
```


## 1.2 C：使用 `scanf`

C 版本的核心輸入：

```c
scanf("%d", &first);
scanf("%d", &second);
```

其中：

```text
%d
```

表示要讀取一個 `int`。

而：

```text
&first
```

表示把 `first` 的記憶體位置交給 `scanf`，讓 `scanf` 可以把讀到的值寫進去。

目前可以先記住：

```text
scanf("%d", &變數);
```

是讀取 `int` 的常見形式。


````python
%%c -n 2

#include <stdio.h>

int main(void) {
    int first, second, sum;

    scanf("%d", &first);
    scanf("%d", &second);

    sum = first + second;

    printf("%d\n", sum);

    return 0;
}
````

## 1.3 手動追蹤 C 程式

假設依序輸入：

```text
5
3
```

程式執行：

```c
scanf("%d", &first);
```

得到：

```text
first = 5
```

接著：

```c
scanf("%d", &second);
```

得到：

```text
second = 3
```

再來：

```c
sum = first + second;
```

所以：

```text
sum = 5 + 3 = 8
```

最後：

```c
printf("%d\n", sum);
```

輸出：

```text
8
```


## 1.4 C++：使用 `cin` 與 `cout`

C++ 可以使用：

```cpp
cin >> first;
cin >> second;
```

輸入資料。

輸出則可以：

```cpp
cout << sum << endl;
```

其中：

```text
>>
```

可以想成「資料流進變數」。

```text
<<
```

可以想成「資料流到輸出畫面」。


````python
%%cpp -n 2

#include <iostream>
using namespace std;

int main(void) {
    int first, second, sum;

    cin >> first;
    cin >> second;

    sum = first + second;

    cout << sum << endl;

    return 0;
}
````

## 1.5 C 與 C++ 對照

| 功能 | C | C++ |
|---|---|---|
| 載入輸入輸出 | `#include <stdio.h>` | `#include <iostream>` |
| 輸入 `int` | `scanf("%d", &x);` | `cin >> x;` |
| 輸出 `int` | `printf("%d\n", x);` | `cout << x << endl;` |

兩個版本的演算法其實完全一樣：

```text
讀 first
讀 second
sum = first + second
輸出 sum
```

差別主要在輸入輸出的語法。


## 1.6 常見錯誤


```c

```

---

### 把比較與指定混淆


```text
=
```

是 **assignment（指定）**，意思是把右邊結果存到左邊。

不是數學式中的「左右兩邊永遠相等」。


# 2. 浮點數輸出格式

原題目給定：

```cpp
double pi = 3.141592653589793;
```

要求分別顯示：

```text
小數點後 2 位
小數點後 4 位
小數點後 8 位
科學記號
```

重點不是改變 `pi` 本身，而是：

> **控制輸出時要顯示多少位。**


## 2.1 C：`printf` 精度格式

C 中：

```c
printf("%.2f\n", pi);
```

可以拆成：

```text
% . 2 f
    ↑
 小數點後 2 位
```

常見形式：

```text
%.2f → 小數點後 2 位
%.4f → 小數點後 4 位
%.8f → 小數點後 8 位
%e   → 科學記號
```


````python
%%c

#include <stdio.h>

int main(void) {
    double pi = 3.141592653589793;

    printf("%.2f\n", pi);
    printf("%.4f\n", pi);
    printf("%.8f\n", pi);
    printf("%e\n", pi);

    return 0;
}
````

## 2.2 預期觀察

大致會看到：

```text
3.14
3.1416
3.14159265
3.141593e+00
```

注意：

```text
3.141592653...
```

顯示成 4 位小數時不是單純砍掉後面，而是會進行 **四捨五入**。

因此：

```text
3.1415...
```

顯示到小數點後 4 位會變成：

```text
3.1416
```


## 2.3 C++：`fixed` 與 `setprecision()`

C++ 需要：

```cpp
#include <iomanip>
```

然後可以使用：

```cpp
fixed
setprecision(...)
scientific
```

例如：

```cpp
cout << fixed << setprecision(2) << pi << endl;
```

意思是：

```text
fixed
→ 使用一般小數形式

setprecision(2)
→ fixed 模式下顯示小數點後 2 位
```


````python
%%cpp

#include <iostream>
#include <iomanip>
using namespace std;

int main(void) {
    double pi = 3.141592653589793;

    cout << fixed << setprecision(2) << pi << endl;
    cout << fixed << setprecision(4) << pi << endl;
    cout << fixed << setprecision(8) << pi << endl;

    cout << scientific << pi << endl;

    return 0;
}
````

## 2.4 一個容易忽略的 C++ 細節：格式狀態會保留

在原程式裡：

```cpp
cout << fixed << setprecision(8) << pi << endl;
cout << scientific << pi << endl;
```

`setprecision(8)` 的設定仍然留在 `cout`。

因此切換成：

```cpp
scientific
```

之後，精度設定仍然會影響輸出。

這也是為什麼 C++ 的 stream manipulators 常被稱作「設定輸出狀態」。


## 2.5 練習

如果：

```cpp
double x = 12.34567;
```

請預測：

```c
printf("%.2f\n", x);
```

以及：

```cpp
cout << fixed << setprecision(3) << x << endl;
```

會顯示什麼。

先判斷要保留幾位，再考慮四捨五入。

# 4. 使用 `sizeof` 查看型別大小

原程式：

```c
sizeof(short)
sizeof(int)
sizeof(long)
sizeof(long long)
sizeof(float)
sizeof(double)
sizeof(char)
```

`sizeof` 會回傳某個型別或物件占用的 **byte 數**。


````python
%%c

#include <stdio.h>

int main(void) {
    printf("short: %zu byte(s)\n", sizeof(short));
    printf("int: %zu byte(s)\n", sizeof(int));
    printf("long: %zu byte(s)\n", sizeof(long));
    printf("long long: %zu byte(s)\n", sizeof(long long));
    printf("float: %zu byte(s)\n", sizeof(float));
    printf("double: %zu byte(s)\n", sizeof(double));
    printf("char: %zu byte(s)\n", sizeof(char));

    return 0;
}
````

## 4.1 為什麼使用 `%zu`？

`sizeof(...)` 的結果型別是：

```c
size_t
```

因此 `printf` 中適合使用：

```text
%zu
```

例如：

```c
printf("%zu\n", sizeof(int));
```

這比把結果硬當成普通 `int` 更正確。


## 4.2 不同電腦結果可能不同

常見平台可能出現：

```text
short      2
int        4
long       4 或 8
long long  8
float      4
double     8
char       1
```

但請以你實際執行：

```c
sizeof(...)
```

得到的結果為準。

特別是：

```text
long
```

在不同作業系統／資料模型下很常不同。


## 4.3 `sizeof(char)` 的特殊性

在 C / C++ 中：

```c
sizeof(char)
```

定義上一定是：

```text
1
```

也就是一個 language-level byte。

現代常見電腦中，一個 byte 通常是 8 bits；  
若要程式化確認每個 byte 有幾個 bits，可以查看：

```c
CHAR_BIT
```

它定義在：

```c

```

這是額外的教學補充，原 notebook 沒有要求這一點。


````python
%%c

#include <stdio.h>
#include <limits.h>

int main(void) {
    printf("CHAR_BIT = %d\n", CHAR_BIT);
    return 0;
}
````

# 5. `float` 與 `double`：為什麼看起來一樣的數會有差？

原程式：

```c
float a = 0.00003f;
double b = 0.00003;

printf("%.20f\n", b - a);
```

表面上：

```text
a = 0.00003
b = 0.00003
```

看起來應該完全相等。

但執行後：

```text
b - a
```

可能不是精確的 0。


````python
%%c

#include <stdio.h>

int main(void) {
    float a = 0.00003f;
    double b = 0.00003;

    printf("%.20f\n", b - a);

    return 0;
}
````

## 5.1 原因：電腦使用二進位近似表示浮點數

很多十進位小數無法用有限個二進位位元精確表示。

因此：

```text
0.00003
```

在記憶體裡通常是某個非常接近它的二進位近似值。

`float` 和 `double` 使用的精度不同，所以：

```text
float 的近似值
```

和：

```text
double 的近似值
```

可能不完全一樣。

所以相減後會看見很小的誤差。


## 5.2 `f` 的作用

原程式：

```c
float a = 0.00003f;
```

最後的：

```text
f
```

表示這個浮點常數是 `float`。

沒有 `f`：

```c
0.00003
```

預設是 `double` 浮點常數。

因此：

```text
0.00003f → float literal
0.00003  → double literal
```


## 5.3 不要直接假設浮點計算完全精確

例如初學者常會期待：

```text
0.1 + 0.2 == 0.3
```

在浮點數世界中一定成立。

實務上，浮點運算可能帶有非常小的表示誤差。

因此之後做數值程式時，常會學到：

```text
比較兩個浮點數是否「足夠接近」
```

而不是永遠使用完全相等。

這是教學延伸；原 notebook 在這裡主要只是示範 `float` 與 `double` 的差異。


# 6. `unsigned int`：從 0 減 2 會發生什麼？

原程式：

```c
unsigned int a = 0;
unsigned int b = a - 2;
```

這裡的重點是：

```text
unsigned
```

表示不使用負整數範圍來表示負值，而是用全部可用位元表示非負整數。


````python
%%c

#include <stdio.h>

int main(void) {
    unsigned int a = 0;
    unsigned int b = a - 2;

    printf("correct unsigned output: %u\n", b);

    return 0;
}
````

## 6.1 unsigned 的回繞概念

如果一個 unsigned 型別能表示：

```text
0 到 MAX
```

那麼從 0 再往下減，結果會依 unsigned 算術規則 **回繞（wrap around）**。

概念上像時鐘：

```text
0
↓ 再減 1
MAX
↓ 再減 1
MAX - 1
```

因此：

```c
0u - 2u
```

會得到接近 unsigned 最大值的數字。

實際是多少，要看該平台的：

```text
unsigned int
```

有多少 bits。


## 6.2 不要把 `%d` 當成 unsigned 的正確輸出格式

原 notebook 同時有：

```c
printf("%%d: %d\n", b);
printf("%%u: %u\n", b);
```

這很適合拿來提醒學生「格式指定符必須和型別相符」。

對：

```c
unsigned int b;
```

正確格式是：

```text
%u
```

而 `%d` 是給 `int` 使用。

因此如果目標是正確、可攜的 C 程式，不應使用 `%d` 來解讀一個 `unsigned int`。

如果想觀察同一組 bits 被 signed 型別如何解讀，應該明確做型別轉換，並理解轉換結果可能與平台規則有關；不要靠錯誤的 `printf` 格式。


````python
%%c

#include <stdio.h>
#include <limits.h>

int main(void) {
    unsigned int a = 0;
    unsigned int b = a - 2;

    printf("UINT_MAX = %u\n", UINT_MAX);
    printf("b        = %u\n", b);

    return 0;
}
````

## 6.3 32-bit unsigned 的常見例子

如果：

```text
unsigned int
```

剛好是 32 bits，那麼範圍是：

```text
0
到
2^32 - 1
```

也就是：

```text
0 到 4294967295
```

所以：

```text
0 - 1 → 4294967295
0 - 2 → 4294967294
```

但不要把這個數字當成所有平台都必然相同。

最安全的做法是用：

```c
UINT_MAX
```

實際查看上限。


# 7. 位元表示：unsigned 與 signed

原 notebook 最後留下的是一段簡短筆記：

```text
# sign + 31(0/1)
# 2bit -> 11 => 3 (unsigned)
# 2bit -> 11 => 1,0,-1 (signed)
```

這是一個片段式提示，沒有完整定義。

下面的內容是 **教學補充**，用 2-bit 小例子把 signed / unsigned 的差異整理完整。


## 7.1 2-bit unsigned

只有 2 bits 時，一共有：

```text
2^2 = 4
```

種 bit patterns：

| bits | unsigned 值 |
|---|---:|
| `00` | 0 |
| `01` | 1 |
| `10` | 2 |
| `11` | 3 |

所以原筆記中的：

```text
2bit -> 11 => 3 (unsigned)
```

可以完整理解為：

```text
11₂ = 3
```

當這兩個 bits 被當成 unsigned 整數時。


## 7.2 2-bit signed：以 two's complement 教學模型說明

現代電腦常以 **two's complement（二補數）** 表示 signed 整數。

用 2-bit 當教學模型：

| bits | signed 值 |
|---|---:|
| `00` | 0 |
| `01` | 1 |
| `10` | -2 |
| `11` | -1 |

所以：

```text
11
```

若當成 2-bit unsigned：

```text
3
```

若當成 2-bit signed two's-complement：

```text
-1
```

這也說明了：

> 同一串 bits 的「數值意義」取決於我們用什麼型別／表示方式解讀。


## 7.3 n-bit 範圍

### unsigned

如果有 `n` bits：

```text
最小值 = 0
最大值 = 2^n - 1
```

### two's-complement signed

教學上可以記成：

```text
最小值 = -2^(n-1)
最大值 =  2^(n-1) - 1
```

例如 8 bits：

```text
unsigned: 0 到 255
signed:  -128 到 127
```

# 12. 最後重點整理

### 基本流程

```text
輸入 → 計算 → 輸出
```

### C

```c
scanf("%d", &x);
printf("%d\n", x);
```

### C++

```cpp
cin >> x;
cout << x << endl;
```

### 浮點格式

C：

```c
%.2f
%.4f
%e
```

C++：

```cpp
fixed
scientific
setprecision(...)
```

### 型別大小

```c
sizeof(type)
```

不要假設所有平台都一樣。

### 浮點數

```text
float / double
```

都可能只能近似表示某些十進位小數。

### unsigned

```text
沒有一般 signed 負數範圍
超過邊界會依 unsigned 規則回繞
```

### 位元

同一串 bits：

```text
11
```

在不同解讀方式下可能代表不同數值。
