````python
# === C / C++ 自動檢查工具：請先執行這個 cell，不要修改 ===
import base64
import json
import os
import shutil
import subprocess
import tempfile
from IPython.core.magic import register_cell_magic

_TEST_CASES = json.loads(
    base64.b64decode("eyJRMSI6W3siaW5wdXQiOiIzIDUgMCAtMiA4XG4iLCJleHBlY3RlZCI6IjMgNSAwIC0yIDhcbiJ9LHsiaW5wdXQiOiIxIDEgMSAxIDFcbiIsImV4cGVjdGVkIjoiMSAxIDEgMSAxXG4ifSx7ImlucHV0IjoiLTUgLTQgLTMgLTIgLTFcbiIsImV4cGVjdGVkIjoiLTUgLTQgLTMgLTIgLTFcbiJ9XSwiUTIiOlt7ImlucHV0IjoiMSAyIDMgNCA1XG4iLCJleHBlY3RlZCI6IjUgNCAzIDIgMVxuIn0seyJpbnB1dCI6IjcgMCAtMiA5IDlcbiIsImV4cGVjdGVkIjoiOSA5IC0yIDAgN1xuIn0seyJpbnB1dCI6Ii0xIC0yIC0zIC00IC01XG4iLCJleHBlY3RlZCI6Ii01IC00IC0zIC0yIC0xXG4ifV0sIlEzIjpbeyJpbnB1dCI6IjEgMiAzIDQgNSA2XG4iLCJleHBlY3RlZCI6IjIxXG4ifSx7ImlucHV0IjoiLTUgLTQgLTMgLTIgLTEgMFxuIiwiZXhwZWN0ZWQiOiItMTVcbiJ9LHsiaW5wdXQiOiIxMCAtMTAgMjAgLTIwIDMwIC0zMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifV0sIlE0IjpbeyJpbnB1dCI6IjEgMiAzIDQgNVxuMTBcbiIsImV4cGVjdGVkIjoiMTEgMTIgMTMgMTQgMTVcbiJ9LHsiaW5wdXQiOiIwIC0xIDUgLTMgOFxuLTJcbiIsImV4cGVjdGVkIjoiLTIgLTMgMyAtNSA2XG4ifSx7ImlucHV0IjoiNyA3IDcgNyA3XG4wXG4iLCJleHBlY3RlZCI6IjcgNyA3IDcgN1xuIn1dLCJRNSI6W3siaW5wdXQiOiIzIDkgNSAxIDhcbiIsImV4cGVjdGVkIjoiOVxuIn0seyJpbnB1dCI6Ii04IC0zIC0xMSAtNCAtOVxuIiwiZXhwZWN0ZWQiOiItM1xuIn0seyJpbnB1dCI6IjUgNSA1IDUgNVxuIiwiZXhwZWN0ZWQiOiI1XG4ifV0sIlE2IjpbeyJpbnB1dCI6IjMgOSA1IDkgMSA4XG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiItOCAtMyAtMTEgLTQgLTMgLTlcbiIsImV4cGVjdGVkIjoiMVxuIn0seyJpbnB1dCI6IjcgNiA1IDQgMyAyXG4iLCJleHBlY3RlZCI6IjBcbiJ9XSwiUTciOlt7ImlucHV0IjoiMSAyIDMgNCA1IDZcbjIgOTlcbiIsImV4cGVjdGVkIjoiMSAyIDk5IDQgNSA2XG4ifSx7ImlucHV0IjoiMCAwIDAgMCAwIDBcbjUgLTdcbiIsImV4cGVjdGVkIjoiMCAwIDAgMCAwIC03XG4ifSx7ImlucHV0IjoiOSA4IDcgNiA1IDRcbjAgMVxuIiwiZXhwZWN0ZWQiOiIxIDggNyA2IDUgNFxuIn1dLCJROCI6W3siaW5wdXQiOiIxMCAyMCAzMCA0MCA1MFxuM1xuIiwiZXhwZWN0ZWQiOiIzMFxuIn0seyJpbnB1dCI6Ii0xIC0yIC0zIC00IC01XG4xXG4iLCJleHBlY3RlZCI6Ii0xXG4ifSx7ImlucHV0IjoiNyA4IDkgMTAgMTFcbjVcbiIsImV4cGVjdGVkIjoiMTFcbiJ9XSwiUTkiOlt7ImlucHV0IjoiMSA1IDIgMFxuIiwiZXhwZWN0ZWQiOiIyMzZcbiJ9LHsiaW5wdXQiOiIzIDMgNCAwXG4iLCJleHBlY3RlZCI6IjI1NVxuIn0seyJpbnB1dCI6IjBcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMTAiOlt7ImlucHV0IjoiMSAyIDMgNCA1IDYgMSAyIDMgNCA1IDZcbiIsImV4cGVjdGVkIjoiMiAyIDIgMiAyIDJcbiJ9LHsiaW5wdXQiOiI2IDYgNiA2IDYgNiA2IDYgNiA2IDYgNlxuIiwiZXhwZWN0ZWQiOiIwIDAgMCAwIDAgMTJcbiJ9LHsiaW5wdXQiOiIxIDEgMSAyIDIgMyA0IDQgNSA2IDYgNlxuIiwiZXhwZWN0ZWQiOiIzIDIgMSAyIDEgM1xuIn1dLCJRMTEiOlt7ImlucHV0IjoiMSAxXG4xIDJcbjIgMlxuMyAzXG42IDZcbjUgNlxuMiA1XG40IDRcbjEgNlxuMyA0XG4iLCJleHBlY3RlZCI6IjEgMSAxIDAgMSAzIDEgMCAwIDEgMVxuIn0seyJpbnB1dCI6IjYgNlxuNiA2XG42IDZcbjYgNlxuNiA2XG42IDZcbjYgNlxuNiA2XG42IDZcbjYgNlxuIiwiZXhwZWN0ZWQiOiIwIDAgMCAwIDAgMCAwIDAgMCAwIDEwXG4ifSx7ImlucHV0IjoiMSA2XG4yIDVcbjMgNFxuNCAzXG41IDJcbjYgMVxuMSAxXG42IDZcbjIgMlxuNSA1XG4iLCJleHBlY3RlZCI6IjEgMCAxIDAgMCA2IDAgMCAxIDAgMVxuIn1dLCJRMTIiOlt7ImlucHV0IjoiMSAyIDMgMiA0IDIgNSAyIDYgMlxuMlxuIiwiZXhwZWN0ZWQiOiI1XG4ifSx7ImlucHV0IjoiMCAwIDAgMCAwIDAgMCAwIDAgMFxuMVxuIiwiZXhwZWN0ZWQiOiIwXG4ifSx7ImlucHV0IjoiLTEgLTIgLTEgLTMgLTEgLTQgLTEgLTUgLTEgLTZcbi0xXG4iLCJleHBlY3RlZCI6IjVcbiJ9XSwiUTEzIjpbeyJpbnB1dCI6IjMgLTEgMCAtNSA4IDIgLTIgNFxuIiwiZXhwZWN0ZWQiOiIzIDAgMCAwIDggMiAwIDRcbiJ9LHsiaW5wdXQiOiItMSAtMiAtMyAtNCAtNSAtNiAtNyAtOFxuIiwiZXhwZWN0ZWQiOiIwIDAgMCAwIDAgMCAwIDBcbiJ9LHsiaW5wdXQiOiIxIDIgMyA0IDUgNiA3IDhcbiIsImV4cGVjdGVkIjoiMSAyIDMgNCA1IDYgNyA4XG4ifV0sIlExNCI6W3siaW5wdXQiOiIxMCAyMCAzMCA0MCA1MFxuMyAxIDUgMFxuIiwiZXhwZWN0ZWQiOiIzMFxuMTBcbjUwXG4ifSx7ImlucHV0IjoiLTUgMCA3IDkgMTJcbjIgNCAwXG4iLCJleHBlY3RlZCI6IjBcbjlcbiJ9LHsiaW5wdXQiOiIxIDEgMiAzIDVcbjUgNCAzIDIgMSAwXG4iLCJleHBlY3RlZCI6IjVcbjNcbjJcbjFcbjFcbiJ9XSwiUTE1IjpbeyJpbnB1dCI6IjEzIDIzIDc4IDc3IDM3IDMwIDQwIDUwIDgzIDk5XG4zMCA3OFxuIiwiZXhwZWN0ZWQiOiI3OCA3NyAzNyAzMCA0MCA1MFxuIn0seyJpbnB1dCI6IjUgMSA5IDIgOCAzIDcgNCA2IDBcbjEwIDIwXG4iLCJleHBlY3RlZCI6Ik5PTkVcbiJ9LHsiaW5wdXQiOiItNSAtMSAwIDQgNyAtMyAyIDkgMSA2XG4tMyAyXG4iLCJleHBlY3RlZCI6Ii0xIDAgLTMgMiAxXG4ifV0sIlExNiI6W3siaW5wdXQiOiIxMyAyMyA3OCA3NyAzNyAzMCA0MCA1MCA4MyA5OVxuMSA0MFxuNzAgODBcbjkwIDEwMFxuMCAwXG4iLCJleHBlY3RlZCI6IjEzIDIzIDM3IDMwIDQwXG43OCA3N1xuOTlcbiJ9LHsiaW5wdXQiOiIxIDIgMyA0IDUgNiA3IDggOSAxMFxuMyAzXG4xMSAyMFxuMCAwXG4iLCJleHBlY3RlZCI6IjNcbk5PTkVcbiJ9LHsiaW5wdXQiOiItNSAtMSAwIDQgNyAtMyAyIDkgMSA2XG4tMyAyXG41IDhcbjAgMFxuIiwiZXhwZWN0ZWQiOiItMSAwIC0zIDIgMVxuNyA2XG4ifV0sIlExNyI6W3siaW5wdXQiOiIxMyAyMyA3OCA3NyAzNyAzMCA0MCA1MCA4MyA5OVxuNDVcbiIsImV4cGVjdGVkIjoiNDBcbiJ9LHsiaW5wdXQiOiIxMCAyMCAzMCA0MCA1MFxuNVxuIiwiZXhwZWN0ZWQiOiJOT05FXG4ifSx7ImlucHV0IjoiLTEwIC00IC03IDIgOVxuLTVcbiIsImV4cGVjdGVkIjoiLTdcbiJ9XSwiUTE4IjpbeyJpbnB1dCI6IjEzIDIzIDc4IDc3IDM3IDMwIDQwIDUwIDgzIDk5XG40NSA4MCAxMiAwXG4iLCJleHBlY3RlZCI6IjQwXG43OFxuTk9ORVxuIn0seyJpbnB1dCI6IjUgMTAgMTUgMjAgMjUgMzAgMzUgNDAgNDUgNTBcbjUgNDkgMyAwXG4iLCJleHBlY3RlZCI6IjVcbjQ1XG5OT05FXG4ifSx7ImlucHV0IjoiMiA0IDYgOCAxMCAxMiAxNCAxNiAxOCAyMFxuMTkgNyAxIDBcbiIsImV4cGVjdGVkIjoiMThcbjZcbk5PTkVcbiJ9XSwiUTE5IjpbeyJpbnB1dCI6IjEwIDIwIDMwIDQwIDUwXG4yNlxuIiwiZXhwZWN0ZWQiOiIzMFxuIn0seyJpbnB1dCI6IjEwIDIwIDMwIDQwIDUwXG4yNVxuIiwiZXhwZWN0ZWQiOiIyMFxuIn0seyJpbnB1dCI6Ii04IC0zIDQgOSAxNVxuMVxuIiwiZXhwZWN0ZWQiOiI0XG4ifV0sIlEyMCI6W3siaW5wdXQiOiIxMCAyMCAzMCA0MCA1MFxuMjYgMjUgNDkgMFxuIiwiZXhwZWN0ZWQiOiIzMFxuMjBcbjUwXG4ifSx7ImlucHV0IjoiMyA4IDE0IDIxIDI5XG4xIDExIDI1IDBcbiIsImV4cGVjdGVkIjoiM1xuOFxuMjFcbiJ9LHsiaW5wdXQiOiI1IDEyIDE5IDI2IDMzXG45IDE2IDMwIDBcbiIsImV4cGVjdGVkIjoiMTJcbjE5XG4zM1xuIn1dLCJRMjEiOlt7ImlucHV0IjoiMSAxMCAxMSAyMCAyMSAzMCAzMSA0MCA5MSAxMDBcbiIsImV4cGVjdGVkIjoiMiAyIDIgMiAwIDAgMCAwIDAgMlxuIn0seyJpbnB1dCI6IjUgNiA3IDggOSAxMCA5OSAxMDAgNTAgNTFcbiIsImV4cGVjdGVkIjoiNiAwIDAgMCAxIDEgMCAwIDAgMlxuIn0seyJpbnB1dCI6IjEwIDIwIDMwIDQwIDUwIDYwIDcwIDgwIDkwIDEwMFxuIiwiZXhwZWN0ZWQiOiIxIDEgMSAxIDEgMSAxIDEgMSAxXG4ifV0sIlEyMiI6W3siaW5wdXQiOiIxIDEwIDExIDIwIDIxIDMwIDMxIDQwIDkxIDEwMFxuIiwiZXhwZWN0ZWQiOiIxMDoqKlxuMjA6KipcbjMwOioqXG40MDoqKlxuNTA6XG42MDpcbjcwOlxuODA6XG45MDpcbjEwMDoqKlxuIn0seyJpbnB1dCI6IjUgNiA3IDggOSAxMCA5OSAxMDAgNTAgNTFcbiIsImV4cGVjdGVkIjoiMTA6KioqKioqXG4yMDpcbjMwOlxuNDA6XG41MDoqXG42MDoqXG43MDpcbjgwOlxuOTA6XG4xMDA6KipcbiJ9LHsiaW5wdXQiOiIxMCAyMCAzMCA0MCA1MCA2MCA3MCA4MCA5MCAxMDBcbiIsImV4cGVjdGVkIjoiMTA6KlxuMjA6KlxuMzA6KlxuNDA6KlxuNTA6KlxuNjA6KlxuNzA6KlxuODA6KlxuOTA6KlxuMTAwOipcbiJ9XSwiUTIzIjpbeyJpbnB1dCI6IjEgMiAyIDMgMyAzIDQgNCA1IDUgNiA3IDggOSAwXG4iLCJleHBlY3RlZCI6IjNcbiJ9LHsiaW5wdXQiOiIxIDEgMiAyIDMgMyA0IDQgNSA1IDYgNiA3IDcgOFxuIiwiZXhwZWN0ZWQiOiI3XG4ifSx7ImlucHV0IjoiOSA5IDkgOSA5IDkgOSA5IDkgOSA5IDkgOSA5IDlcbiIsImV4cGVjdGVkIjoiOVxuIn1dLCJRMjQiOlt7ImlucHV0IjoiMSAyIDMgNCA1IDYgNyA4IDkgMTBcbiIsImV4cGVjdGVkIjoiMTAgNVxuIn0seyJpbnB1dCI6Ii0xMCAtOSAtOCAtNyAtNiAtNSAtNCAtMyAtMiAtMVxuIiwiZXhwZWN0ZWQiOiItMSAtNVxuIn0seyJpbnB1dCI6IjEwIC0xMCAyMCAtMjAgMzAgLTMwIDQwIC00MCA1MCAtNTBcbiIsImV4cGVjdGVkIjoiNTAgMFxuIn1dLCJRMjUiOlt7ImlucHV0IjoiMTMgMjMgNzggNzcgMzcgMzAgNDAgNTAgODMgOTlcbjMwIDc4XG4iLCJleHBlY3RlZCI6IjZcbiJ9LHsiaW5wdXQiOiIxIDIgMyA0IDUgNiA3IDggOSAxMFxuMTEgMjBcbiIsImV4cGVjdGVkIjoiMFxuIn0seyJpbnB1dCI6Ii01IC0xIDAgNCA3IC0zIDIgOSAxIDZcbi0zIDJcbiIsImV4cGVjdGVkIjoiNVxuIn1dLCJRMjYiOlt7ImlucHV0IjoiMSAxIDIgMiAzIDMgNCA0IDUgNSA1IDVcbiIsImV4cGVjdGVkIjoiNVxuIn0seyJpbnB1dCI6IjAgMCAwIDAgMCAwIDAgMCAwIDAgMCAwXG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiIwIDEgMiAzIDQgNSA2IDcgOCA5IDAgOVxuIiwiZXhwZWN0ZWQiOiIxMFxuIn1dLCJRMjciOlt7ImlucHV0IjoiOSA0IDYgMiAxXG4iLCJleHBlY3RlZCI6IjQgNiAyIDEgOVxuIn0seyJpbnB1dCI6IjEgMiAzIDQgNVxuIiwiZXhwZWN0ZWQiOiIxIDIgMyA0IDVcbiJ9LHsiaW5wdXQiOiI1IDUgMiAzIDhcbiIsImV4cGVjdGVkIjoiNSAyIDMgNSA4XG4ifV0sIlEyOCI6W3siaW5wdXQiOiI5IDQgNiAyIDFcbiIsImV4cGVjdGVkIjoiNCAyIDEgNiA5XG4ifSx7ImlucHV0IjoiNSAxIDQgMiA4XG4iLCJleHBlY3RlZCI6IjEgMiA0IDUgOFxuIn0seyJpbnB1dCI6IjMgMyAyIDEgM1xuIiwiZXhwZWN0ZWQiOiIyIDEgMyAzIDNcbiJ9XSwiUTI5IjpbeyJpbnB1dCI6IjkgNCA2IDIgMVxuIiwiZXhwZWN0ZWQiOiIxIDIgNCA2IDlcbiJ9LHsiaW5wdXQiOiI1IDUgMiAzIDhcbiIsImV4cGVjdGVkIjoiMiAzIDUgNSA4XG4ifSx7ImlucHV0IjoiLTEgLTUgMyAwIDJcbiIsImV4cGVjdGVkIjoiLTUgLTEgMCAyIDNcbiJ9XSwiUTMwIjpbeyJpbnB1dCI6IjkgNCA2IDIgMVxuIiwiZXhwZWN0ZWQiOiI5IDYgNCAyIDFcbiJ9LHsiaW5wdXQiOiI1IDUgMiAzIDhcbiIsImV4cGVjdGVkIjoiOCA1IDUgMyAyXG4ifSx7ImlucHV0IjoiLTEgLTUgMyAwIDJcbiIsImV4cGVjdGVkIjoiMyAyIDAgLTEgLTVcbiJ9XSwiUTMxIjpbeyJpbnB1dCI6IjkgNCA2IDIgMVxuIiwiZXhwZWN0ZWQiOiIxIDlcbiJ9LHsiaW5wdXQiOiItNSAtMiAtOSAtMSAtN1xuIiwiZXhwZWN0ZWQiOiItOSAtMVxuIn0seyJpbnB1dCI6IjMgMyAzIDMgM1xuIiwiZXhwZWN0ZWQiOiIzIDNcbiJ9XSwiUTMyIjpbeyJpbnB1dCI6IjkgNCA2IDIgMVxuIiwiZXhwZWN0ZWQiOiI0XG4ifSx7ImlucHV0IjoiNSA1IDIgMyA4XG4iLCJleHBlY3RlZCI6IjVcbiJ9LHsiaW5wdXQiOiItMSAtNSAzIDAgMlxuIiwiZXhwZWN0ZWQiOiIwXG4ifV0sIlEzMyI6W3siaW5wdXQiOiI1XG41IDQgMyAyIDFcbiIsImV4cGVjdGVkIjoiMTBcbiJ9LHsiaW5wdXQiOiIxXG40MlxuIiwiZXhwZWN0ZWQiOiIwXG4ifSx7ImlucHV0IjoiOFxuMyAxIDQgMSA1IDkgMiA2XG4iLCJleHBlY3RlZCI6IjI4XG4ifV0sIlEzNCI6W3siaW5wdXQiOiI1XG41IDQgMyAyIDFcbiIsImV4cGVjdGVkIjoiMTBcbiJ9LHsiaW5wdXQiOiI1XG4xIDIgMyA0IDVcbiIsImV4cGVjdGVkIjoiMFxuIn0seyJpbnB1dCI6IjZcbjMgMSAyIDYgNSA0XG4iLCJleHBlY3RlZCI6IjVcbiJ9XSwiUTM1IjpbeyJpbnB1dCI6IjVcbjEgMiAzIDQgNVxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiNVxuNSA0IDMgMiAxXG4iLCJleHBlY3RlZCI6IjRcbiJ9LHsiaW5wdXQiOiI2XG4xIDMgMiA0IDUgNlxuIiwiZXhwZWN0ZWQiOiIyXG4ifV0sIlEzNiI6W3siaW5wdXQiOiI5IDQgNiAyIDFcbiIsImV4cGVjdGVkIjoiNCA2IDIgMSA5XG40IDIgMSA2IDlcbjIgMSA0IDYgOVxuMSAyIDQgNiA5XG4ifSx7ImlucHV0IjoiMSAyIDMgNCA1XG4iLCJleHBlY3RlZCI6IjEgMiAzIDQgNVxuMSAyIDMgNCA1XG4xIDIgMyA0IDVcbjEgMiAzIDQgNVxuIn0seyJpbnB1dCI6IjUgMSA0IDIgOFxuIiwiZXhwZWN0ZWQiOiIxIDQgMiA1IDhcbjEgMiA0IDUgOFxuMSAyIDQgNSA4XG4xIDIgNCA1IDhcbiJ9XSwiUTM3IjpbeyJpbnB1dCI6IjEgMiAzIDQgNSA2XG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiIxIDIgMiAzIDQgNFxuIiwiZXhwZWN0ZWQiOiIxXG4ifSx7ImlucHV0IjoiMSAzIDIgNCA1IDZcbiIsImV4cGVjdGVkIjoiMFxuIn1dLCJRMzgiOlt7ImlucHV0IjoiNSAtMSA1IDIgLTMgMiAwIDhcbiIsImV4cGVjdGVkIjoiLTMgLTEgMCAyIDIgNSA1IDhcbiJ9LHsiaW5wdXQiOiI4IDcgNiA1IDQgMyAyIDFcbiIsImV4cGVjdGVkIjoiMSAyIDMgNCA1IDYgNyA4XG4ifSx7ImlucHV0IjoiLTQgLTQgLTEgLTkgMCAzIDMgMlxuIiwiZXhwZWN0ZWQiOiItOSAtNCAtNCAtMSAwIDIgMyAzXG4ifV0sIlEzOSI6W3siaW5wdXQiOiIxIDIgMyA0IDUgNlxuIiwiZXhwZWN0ZWQiOiIxIDIgM1xuNCA1IDZcbiJ9LHsiaW5wdXQiOiItMSAwIDEgMiAzIDRcbiIsImV4cGVjdGVkIjoiLTEgMCAxXG4yIDMgNFxuIn0seyJpbnB1dCI6IjkgOSA4IDggNyA3XG4iLCJleHBlY3RlZCI6IjkgOSA4XG44IDcgN1xuIn1dLCJRNDAiOlt7ImlucHV0IjoiMSAyIDMgNCA1IDZcbjEgMlxuIiwiZXhwZWN0ZWQiOiI2XG4ifSx7ImlucHV0IjoiLTEgMCAxIDIgMyA0XG4wIDFcbiIsImV4cGVjdGVkIjoiMFxuIn0seyJpbnB1dCI6IjkgOCA3IDYgNSA0XG4xIDBcbiIsImV4cGVjdGVkIjoiNlxuIn1dLCJRNDEiOlt7ImlucHV0IjoiMSAyIDMgNCA1IDYgNyA4IDlcbiIsImV4cGVjdGVkIjoiNDVcbiJ9LHsiaW5wdXQiOiIwIDAgMCAwIDAgMCAwIDAgMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifSx7ImlucHV0IjoiLTQgLTMgLTIgLTEgMCAxIDIgMyA0XG4iLCJleHBlY3RlZCI6IjBcbiJ9XSwiUTQyIjpbeyJpbnB1dCI6IjEgMiAzIDQgNSA2IDcgOCA5XG4iLCJleHBlY3RlZCI6IjYgMTUgMjRcbiJ9LHsiaW5wdXQiOiIwIDAgMCAwIDAgMCAwIDAgMFxuIiwiZXhwZWN0ZWQiOiIwIDAgMFxuIn0seyJpbnB1dCI6Ii00IC0zIC0yIC0xIDAgMSAyIDMgNFxuIiwiZXhwZWN0ZWQiOiItOSAwIDlcbiJ9XSwiUTQzIjpbeyJpbnB1dCI6IjEgMiAzIDQgNSA2IDcgOCA5XG4iLCJleHBlY3RlZCI6IjEyIDE1IDE4XG4ifSx7ImlucHV0IjoiMCAwIDAgMCAwIDAgMCAwIDBcbiIsImV4cGVjdGVkIjoiMCAwIDBcbiJ9LHsiaW5wdXQiOiItNCAtMyAtMiAtMSAwIDEgMiAzIDRcbiIsImV4cGVjdGVkIjoiLTMgMCAzXG4ifV0sIlE0NCI6W3siaW5wdXQiOiIxIDIgMyA0IDUgNiA3IDggOVxuIiwiZXhwZWN0ZWQiOiIxIDQgN1xuMiA1IDhcbjMgNiA5XG4ifSx7ImlucHV0IjoiMCAwIDAgMCAwIDAgMCAwIDBcbiIsImV4cGVjdGVkIjoiMCAwIDBcbjAgMCAwXG4wIDAgMFxuIn0seyJpbnB1dCI6Ii00IC0zIC0yIC0xIDAgMSAyIDMgNFxuIiwiZXhwZWN0ZWQiOiItNCAtMSAyXG4tMyAwIDNcbi0yIDEgNFxuIn1dLCJRNDUiOlt7ImlucHV0IjoiMVxuIiwiZXhwZWN0ZWQiOiIxIDIgM1xuNCA1IDZcbjcgOCA5XG4ifSx7ImlucHV0IjoiMTBcbiIsImV4cGVjdGVkIjoiMTAgMTEgMTJcbjEzIDE0IDE1XG4xNiAxNyAxOFxuIn0seyJpbnB1dCI6Ii0zXG4iLCJleHBlY3RlZCI6Ii0zIC0yIC0xXG4wIDEgMlxuMyA0IDVcbiJ9XSwiUTQ2IjpbeyJpbnB1dCI6IjAgMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifSx7ImlucHV0IjoiMSAyXG4iLCJleHBlY3RlZCI6IjVcbiJ9LHsiaW5wdXQiOiIyIDFcbiIsImV4cGVjdGVkIjoiN1xuIn1dLCJRNDciOlt7ImlucHV0IjoiMFxuIiwiZXhwZWN0ZWQiOiIwIDBcbiJ9LHsiaW5wdXQiOiI1XG4iLCJleHBlY3RlZCI6IjEgMlxuIn0seyJpbnB1dCI6IjhcbiIsImV4cGVjdGVkIjoiMiAyXG4ifV0sIlE0OCI6W3siaW5wdXQiOiIyIDQgMSAzXG4iLCJleHBlY3RlZCI6IjdcbiJ9LHsiaW5wdXQiOiIzIDUgMiA0XG4iLCJleHBlY3RlZCI6IjE0XG4ifSx7ImlucHV0IjoiNCAyIDMgMVxuIiwiZXhwZWN0ZWQiOiI3XG4ifV0sIlE0OSI6W3siaW5wdXQiOiIxIDIgMyA0IDUgNiA3IDggOVxuIiwiZXhwZWN0ZWQiOiIxIDIgMyA0IDUgNiA3IDggOVxuIn0seyJpbnB1dCI6Ii0xIDAgMSAyIDMgNCA1IDYgN1xuIiwiZXhwZWN0ZWQiOiItMSAwIDEgMiAzIDQgNSA2IDdcbiJ9LHsiaW5wdXQiOiI5IDggNyA2IDUgNCAzIDIgMVxuIiwiZXhwZWN0ZWQiOiI5IDggNyA2IDUgNCAzIDIgMVxuIn1dLCJRNTAiOlt7ImlucHV0IjoiMSAyIDMgNCA1IDYgNyA4IDlcbjAgOFxuIiwiZXhwZWN0ZWQiOiI5IDIgM1xuNCA1IDZcbjcgOCAxXG4ifSx7ImlucHV0IjoiOSA4IDcgNiA1IDQgMyAyIDFcbjEgN1xuIiwiZXhwZWN0ZWQiOiI5IDIgN1xuNiA1IDRcbjMgOCAxXG4ifSx7ImlucHV0IjoiLTEgMCAxIDIgMyA0IDUgNiA3XG40IDRcbiIsImV4cGVjdGVkIjoiLTEgMCAxXG4yIDMgNFxuNSA2IDdcbiJ9XX0=").decode("utf-8")
)

_ANSWERS = {}
_RESULTS = {}

def _normalise_output(text):
    lines = [line.rstrip() for line in text.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _find_compiler(language):
    if language == "c":
        names = ["gcc", "clang", "cc"]
    else:
        names = ["g++", "clang++", "c++"]

    for name in names:
        path = shutil.which(name)
        if path:
            return path
    return None

def _compile_and_run(source, input_text, language):
    compiler = _find_compiler(language)

    if compiler is None:
        if language == "c":
            return False, "", "找不到 C 編譯器（gcc / clang / cc）。"
        return False, "", "找不到 C++ 編譯器（g++ / clang++ / c++）。"

    with tempfile.TemporaryDirectory() as td:
        if language == "c":
            src = os.path.join(td, "answer.c")
            compile_cmd = [compiler, "-std=c11", src, "-o", os.path.join(td, "answer")]
        else:
            src = os.path.join(td, "answer.cpp")
            compile_cmd = [compiler, "-std=c++17", src, "-o", os.path.join(td, "answer")]

        exe = os.path.join(td, "answer")

        with open(src, "w", encoding="utf-8") as f:
            f.write(source)

        compile_result = subprocess.run(
            compile_cmd,
            capture_output=True,
            text=True,
        )

        if compile_result.returncode != 0:
            return False, "", compile_result.stderr

        try:
            run_result = subprocess.run(
                [exe],
                input=input_text,
                capture_output=True,
                text=True,
                timeout=5,
            )
        except subprocess.TimeoutExpired:
            return False, "", "程式執行超過 5 秒，請檢查是否有無限迴圈。"

        if run_result.returncode != 0:
            err = run_result.stderr or f"程式結束碼：{run_result.returncode}"
            return False, run_result.stdout, err

        return True, run_result.stdout, ""

def _save_answer(qid, language, cell):
    qid = qid.strip().upper()
    if qid not in _TEST_CASES:
        print(f"❌ 未知題號：{qid}")
        return

    _ANSWERS[qid] = {"language": language, "source": cell}
    label = "C" if language == "c" else "C++"
    print(f"已儲存 {qid} 的 {label} 作答。請執行下方檢查 cell。")

@register_cell_magic
def c_answer(line, cell):
    _save_answer(line, "c", cell)

@register_cell_magic
def cpp_answer(line, cell):
    _save_answer(line, "cpp", cell)

def _run_check(qid):
    qid = qid.upper()

    if qid not in _ANSWERS:
        print(f"❌ {qid}：尚未執行 C 或 C++ 作答 cell")
        _RESULTS[qid] = False
        return

    answer = _ANSWERS[qid]
    language = answer["language"]
    source = answer["source"]
    cases = _TEST_CASES[qid]
    passed = 0

    for case_number, case in enumerate(cases, start=1):
        ok_run, stdout, error = _compile_and_run(
            source,
            case["input"],
            language,
        )

        if not ok_run:
            label = "C" if language == "c" else "C++"
            print(f"❌ {qid}：{label} 編譯或執行失敗")
            print(error)
            _RESULTS[qid] = False
            return

        actual = _normalise_output(stdout)
        expected = _normalise_output(case["expected"])

        if actual == expected:
            passed += 1
        else:
            print(f"❌ {qid} 測試 {case_number} 未通過")
            print("測試輸入：")
            print(case["input"], end="")
            print("你的輸出：")
            print(stdout if stdout else "<沒有輸出>")
            print("預期輸出：")
            print(case["expected"], end="")
            print()

    success = passed == len(cases)
    _RESULTS[qid] = success
    label = "C" if language == "c" else "C++"

    if success:
        print(f"✅ {qid} 全部通過（{passed} / {len(cases)}，{label}）")
    else:
        print(f"❌ {qid}：通過 {passed} / {len(cases)} 個測試（{label}）")

def _final_report():
    passed = sum(_RESULTS.get(f"Q{i}", False) for i in range(1, 51))
    untouched = [f"Q{i}" for i in range(1, 51) if f"Q{i}" not in _RESULTS]

    print(f"總成績：{passed} / 50 題通過")
    if untouched:
        print("尚未執行檢查：" + ", ".join(untouched))

print("C / C++ 自動檢查工具已載入。")
print("每題可選 C 或 C++；最後執行的作答 cell 會作為該題答案。")
print("作答後不需要手動輸入測試資料。")
````
